### Step 2 — Create Rating Category

```python
def rating_category(rating):
    if rating >= 8:
        return 'Excellent'
    elif rating >= 7:
        return 'Good'
    elif rating >= 6:
        return 'Average'
    else:
        return 'Below Average'

df['Rating_Category'] = df['Rating'].apply(rating_category)
df[['Movie', 'Rating', 'Rating_Category']].head(10)
```

### Step 3 — Create Runtime Category

```python
def runtime_category(runtime):
    if runtime < 120:
        return 'Short'
    elif runtime <= 150:
        return 'Medium'
    else:
        return 'Long'

df['Runtime_Category'] = df['Runtime'].apply(runtime_category)
df[['Movie', 'Runtime', 'Runtime_Category']].head(10)
```

- **`if / elif / else`** → creates categories using custom conditions.
- **`apply()`** → applies the categorization logic sequentially over the Runtime column.

**Purpose**

## Topic: Movie Recommendation System

### Sampath will build a content-based movie recommendation system using the movie's genre and overview.

---

### Step 1 — Prepare Recommendation Data

```python
df['Recommendation_Text'] = (
    df['Genre'].fillna('') + ' ' +
    df['Overview'].fillna('')
)
df[['Movie', 'Recommendation_Text']].head()
```

- **`Recommendation_Text`** → new column containing information used for recommendations.
- **`fillna('')`** → replaces missing values with empty text.

**Purpose**
Combines the movie's genre and overview into one text field for finding similar movies.

### Step 2 — Convert Text into Numerical Features

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf = TfidfVectorizer(stop_words='english')
tfidf_matrix = tfidf.fit_transform(df['Recommendation_Text'])
print("TF-IDF Matrix Shape:", tfidf_matrix.shape)
```

- **`TfidfVectorizer()`** → converts text into numerical features.
- **`fit_transform()`** → learns the important words and transforms the text into numbers.

**Purpose**
Converts movie descriptions into numerical data so that similarities between movies can be calculated.

### Step 3 — Calculate Movie Similarity

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity_matrix = cosine_similarity(tfidf_matrix)
print("Similarity Matrix Shape:", similarity_matrix.shape)
```

- **`cosine_similarity()`** → measures how similar two movie descriptions are.
- **`similarity_matrix`** → stores the similarity scores between movies.

**Purpose**
Calculates how similar every movie is to every other movie.

### Step 4 — Create Recommendation Function

```python
def recommend_movies(movie_name, n=5):
    matches = df[df['Movie'].str.lower() == movie_name.lower()]
    if matches.empty:
        return "Movie not found"
    movie_index = matches.index[0]
    similarity_scores = list(enumerate(similarity_matrix[movie_index]))
    similarity_scores = sorted(
        similarity_scores,
        key=lambda x: x[1],
        reverse=True
    )
    movie_indices = [i[0] for i in similarity_scores[1:n+1]]
    return df.iloc[movie_indices][
        ['Movie', 'Genre', 'Rating']
    ]
```

- **`def`** → creates a recommendation function.
- **`similarity_scores`** → stores similarity values.
- **`sorted()`** → arranges movies by similarity.
- **`reverse=True`** → puts the most similar movies first.
- **`iloc[]`** → selects movies using their index.

**Purpose**
Creates a function that finds and returns the most similar movies for a given movie.

### Step 5 — Test the Recommendation System

```python
recommend_movies("Bahubali: The Beginning", 5)
```

- **`recommend_movies()`** → gives recommendations for the selected movie.

**Purpose**
Tests the recommendation system by finding five movies similar to the selected movie.

### Step 6 — Display Recommended Movies

```python
recommendations = recommend_movies("Bahubali: The Beginning", 5)
print("Recommended Movies:")
display(recommendations)
```

- **`recommendations`** → stores the movies suggested by the system.

**Purpose**
Displays the final movie recommendations with their genre and ratings.

- **`lineplot()`** → charts serial trajectory changes over discrete temporal indexes.
- **`marker='o'`** → places solid anchor dots for each timeline category value.

**Purpose**

**Purpose**
Examines whether movie runtime has a visible relationship with movie rating.

### Step 9 — Filter the Dataset

```python
def filter_movies(genre):
    if genre == 'All':
        filtered_df = df
    else:
        filtered_df = df[
            df['Genre'].str.contains(
                genre,
                case=False,
                na=False
            )
        ]
    fig = px.scatter(
        filtered_df,
        x='Runtime',
        y='Rating',
        hover_name='Movie',
        title=f'Runtime vs Rating - {genre}'
    )
    fig.show()
widgets.interact(
    filter_movies,
    genre=genre_dropdown
)
```
